```python
import pandas as pd
```

```python
def load_and_label(path, label):
    df = pd.read_csv(path)
    df['type'] = label
    return df

df_r_cnn_l = load_and_label("output/layertuning/hyperparameter_tuning_kerastuner_random_layertuning/cnn/final_test_evaluation.csv", "random_layertuning")
df_r_mlp_l = load_and_label("output/layertuning/hyperparameter_tuning_kerastuner_random_layertuning/mlp/final_test_evaluation.csv", "random_layertuning")
df_b_cnn_l = load_and_label("output/layertuning/hyperparameter_tuning_kerastuner_bayesian_layertuning/cnn/final_test_evaluation.csv", "bayesian_layertuning")
df_b_mlp_l = load_and_label("output/layertuning/hyperparameter_tuning_kerastuner_bayesian_layertuning/mlp/final_test_evaluation.csv", "bayesian_layertuning")

df_r_cnn = load_and_label("output/hyperparameter_tuning_kerastuner_random/cnn/final_test_evaluation.csv", "random")
df_r_mlp = load_and_label("output/hyperparameter_tuning_kerastuner_random/mlp/final_test_evaluation.csv", "random")
df_b_cnn = load_and_label("output/hyperparameter_tuning_kerastuner_bayesian/cnn/final_test_evaluation.csv", "bayesian")
df_b_mlp = load_and_label("output/hyperparameter_tuning_kerastuner_bayesian/mlp/final_test_evaluation.csv", "bayesian")

df_combined = pd.concat([df_r_cnn_l, df_r_mlp_l, df_b_cnn_l, df_b_mlp_l, df_r_cnn, df_r_mlp, df_b_cnn, df_b_mlp], ignore_index=True)
df_combined = df_combined[['type'] + [col for col in df_combined.columns if col != 'type']]
display(df_combined.sort_values('test_accuracy', ascending=False))
```

```python
# Load the consolidated results CSV
consolidated_path = "output/TomSplit/all_evaluations_consolidated.csv"
df = pd.read_csv(consolidated_path)
df["f1"] = 2 * (df["precision"] * df["recall"]) / (df["precision"] + df["recall"])
cols = list(df.columns)
recall_idx = cols.index("recall")
# move f1 to the left
cols.remove("f1")
cols.insert(recall_idx + 1, "f1")
df = df[cols]

# Show the DataFrame with color highlighting for accuracy, precision, and recall
styled_df = df.style.background_gradient(subset=[
    col for col in df.columns if any(metric in col for metric in ["accuracy", "precision", "recall", "f1"])
], cmap="coolwarm")
display(styled_df)
```
